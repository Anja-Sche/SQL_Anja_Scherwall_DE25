# Create a duckdb database in Python
- use with duckdb.connect() to connect to a duckdb database
- if the database file doesn't exist, it will create one

```python
import duckdb

with duckdb.connect("databases/hotels.duckdb") as conn:
    conn.sql("DROP TABLE IF EXISTS bookings;")
    conn.sql(
        """
        CREATE TABLE IF NOT EXISTS bookings (
            date DATE, 
            room_type STRING,
            price FLOAT,
            breakfast BOOLEAN,
            name STRING,
            email STRING,
            guests UINT8
        )
    """
    )
```
